# GG-PA quickstart

This notebook runs the two CPU quickstarts of [GG-PA](https://github.com/weizhouwangw/GG-PA) (Generative Gibbs for Physics-Aware Sampling, [arXiv 2605.10642](https://arxiv.org/abs/2605.10642)) and shows their figures.

- **Linear inverse problem.** Split Gibbs sampling with a pretrained prior, with and without the covariance correction that makes it exact.
- **Double well in an environment.** A diffusion model trained on an isolated double well is coupled to a harmonic environment at sampling time, without retraining.

Choose **Runtime → Run all**. No GPU is needed. The settings and the formulas are in [quickstart/README.md](https://github.com/weizhouwangw/GG-PA/blob/main/quickstart/README.md).

## 3. Look at the figures

```python
from pathlib import Path
from IPython.display import Image, display

run = sorted(Path("runs").glob("quickstart_*"))[-1]
for figure in sorted((run / "figures").glob("*.png")):
    print(figure.name)
    display(Image(filename=str(figure)))
```

The double-well figure shows the distribution of the clean coordinate from the model alone, from GG-PA and from GG-PA with replica exchange, against the exact marginal. The inverse-problem figure compares the corrected and the uncorrected split Gibbs sampler with the exact posterior.
